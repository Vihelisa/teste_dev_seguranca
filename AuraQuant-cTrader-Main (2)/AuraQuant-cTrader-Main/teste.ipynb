```python
from decimal import Decimal

def test_lot_calculation():
    """Testa cálculo de tamanho de lote com Decimal"""
    
    print("="*60)
    print("TESTE: Cálculo de Tamanho de Lote")
    print("="*60)
    
    # Dados de entrada
    fixed_lot_size = Decimal('0.10')
    account_balance = Decimal('10000.00')
    risk_percent = Decimal('2.00')
    entry_price = Decimal('1.10000')
    sl_price = Decimal('1.09500')
    
    print(f"\n📊 Dados de Entrada:")
    print(f"  Lote Fixo: {fixed_lot_size}")
    print(f"  Saldo: ${account_balance}")
    print(f"  Risco: {risk_percent}%")
    print(f"  Preço Entrada: {entry_price}")
    print(f"  Stop Loss: {sl_price}")
    
    # Cálculo DYNAMIC
    print(f"\n💰 Cálculo DYNAMIC:")
    
    # Risco monetário
    risk_amount_money = account_balance * (risk_percent / Decimal('100.0'))
    print(f"  1. Risco Monetário: ${risk_amount_money}")
    
    # Distância SL
    sl_distance_pips = abs(entry_price - sl_price)
    print(f"  2. Distância SL: {sl_distance_pips} pips")
    
    # Valor por lote
    value_per_lot = Decimal('10.0')
    print(f"  3. Valor por lote: ${value_per_lot} por pip")
    
    # Cálculo de lote
    if sl_distance_pips > 0:
        calculated_lot_size = risk_amount_money / (sl_distance_pips * value_per_lot)
    else:
        calculated_lot_size = risk_amount_money / Decimal('1000.0')
    
    print(f"  4. Lote Calculado: {calculated_lot_size}")
    
    # Arredondar
    calculated_lot_size = round(calculated_lot_size, 2)
    print(f"  5. Lote Arredondado: {calculated_lot_size}")
    
    # Aplicar limites
    MIN_LOT = Decimal('0.01')
    MAX_LOT = Decimal('100.0')
    final_lot_size = max(MIN_LOT, min(calculated_lot_size, MAX_LOT))
    print(f"  6. Lote Final (com limites): {final_lot_size}")
    
    # Validações
    print(f"\n✅ Validações:")
    assert isinstance(final_lot_size, Decimal), "❌ Resultado não é Decimal!"
    assert final_lot_size >= MIN_LOT, f"❌ Lote menor que mínimo: {final_lot_size}"
    assert final_lot_size <= MAX_LOT, f"❌ Lote maior que máximo: {final_lot_size}"
    print(f"  ✓ Tipo correto: Decimal")
    print(f"  ✓ Dentro dos limites: {MIN_LOT} <= {final_lot_size} <= {MAX_LOT}")
    
    return final_lot_size


def test_precision():
    """Testa precisão de Decimal vs Float"""
    
    print("\n" + "="*60)
    print("TESTE: Precisão Decimal vs Float")
    print("="*60)
    
    # Problema clássico do Float
    float_result = 0.1 + 0.2
    decimal_result = Decimal('0.1') + Decimal('0.2')
    
    print(f"\n🔬 0.1 + 0.2:")
    print(f"  Float:   {float_result}")
    print(f"  Float == 0.3? {float_result == 0.3}")
    print(f"  Decimal: {decimal_result}")
    print(f"  Decimal == 0.3? {decimal_result == Decimal('0.3')}")
    
    # Teste com percentuais
    print(f"\n📊 Cálculo de Risco:")
    balance_float = 10000.0 * (2.5 / 100.0)
    balance_decimal = Decimal('10000.00') * (Decimal('2.5') / Decimal('100.0'))
    
    print(f"  Float:   ${balance_float}")
    print(f"  Decimal: ${balance_decimal}")
    print(f"  Diferença: ${abs(Decimal(str(balance_float)) - balance_decimal)}")


def test_edge_cases():
    """Testa casos extremos"""
    
    print("\n" + "="*60)
    print("TESTE: Casos Extremos")
    print("="*60)
    
    # Caso 1: SL distance zero
    print("\n🔍 Caso 1: Distância SL = 0")
    risk_money = Decimal('200.00')
    sl_distance = Decimal('0')
    
    if sl_distance > 0:
        lot = risk_money / (sl_distance * Decimal('10.0'))
    else:
        lot = risk_money / Decimal('1000.0')
    
    print(f"  Risco: ${risk_money}")
    print(f"  SL Distance: {sl_distance}")
    print(f"  Lote (fallback): {lot}")
    assert lot == Decimal('0.20'), "❌ Fallback incorreto"
    print(f"  ✓ Fallback funcionando")
    
    # Caso 2: Saldo zero
    print("\n🔍 Caso 2: Saldo = 0")
    balance = Decimal('0')
    fixed_lot = Decimal('0.10')
    
    if balance > 0:
        final_lot = Decimal('999')  # Não deve chegar aqui
    else:
        final_lot = fixed_lot
    
    print(f"  Saldo: ${balance}")
    print(f"  Lote usado: {final_lot} (fixo)")
    assert final_lot == fixed_lot, "❌ Não usou lote fixo"
    print(f"  ✓ Usando lote fixo corretamente")
    
    # Caso 3: Lote calculado muito alto
    print("\n🔍 Caso 3: Lote Calculado > MAX")
    calculated = Decimal('150.00')
    MAX_LOT = Decimal('100.0')
    MIN_LOT = Decimal('0.01')
    
    final = max(MIN_LOT, min(calculated, MAX_LOT))
    print(f"  Calculado: {calculated}")
    print(f"  MAX: {MAX_LOT}")
    print(f"  Final: {final}")
    assert final == MAX_LOT, "❌ Não aplicou limite máximo"
    print(f"  ✓ Limite máximo aplicado")
    
    # Caso 4: Lote calculado muito baixo
    print("\n🔍 Caso 4: Lote Calculado < MIN")
    calculated = Decimal('0.005')
    
    final = max(MIN_LOT, min(calculated, MAX_LOT))
    print(f"  Calculado: {calculated}")
    print(f"  MIN: {MIN_LOT}")
    print(f"  Final: {final}")
    assert final == MIN_LOT, "❌ Não aplicou limite mínimo"
    print(f"  ✓ Limite mínimo aplicado")


if __name__ == '__main__':
    try:
        # Rodar testes
        test_precision()
        lot = test_lot_calculation()
        test_edge_cases()
        
        print("\n" + "="*60)
        print("✅ TODOS OS TESTES PASSARAM!")
        print("="*60)
        print(f"\nLote final calculado: {lot} lotes")
        
    except AssertionError as e:
        print(f"\n❌ TESTE FALHOU: {e}")
    except Exception as e:
        print(f"\n❌ ERRO: {e}")
        import traceback
        traceback.print_exc()
```
